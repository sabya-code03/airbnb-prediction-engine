```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset from the data directory
df = pd.read_csv('../data/raw.csv', low_memory=False, encoding='latin1')

# Display the first 5 rows to understand feature formatting
display(df.head())

# Check dataset dimensions, column data types, and non-null counts
df.info()

# Generate summary statistics for numerical columns like price
display(df.describe())
```

```python
# 1. Visualize the skewed raw price
plt.figure(figsize=(10, 4))
sns.histplot(df['price'], bins=50)
plt.title('Raw Price Distribution (Notice the extreme skew)')
plt.show()

# 2. Filter outliers: Keep realistic prices ($10 to $1500) and realistic minimum nights
clean_df = df[(df['price'] >= 10) & (df['price'] <= 1500) & (df['minimum_nights'] <= 30)].copy()

# 3. Drop columns with excessive missing data or irrelevant identifiers
cols_to_drop = ['listing_id', 'name', 'host_id', 'district', 'host_profile_pic']
clean_df = clean_df.drop(columns=cols_to_drop, errors='ignore')

# 4. Visualize the cleaned, normalized price distribution
plt.figure(figsize=(10, 4))
sns.histplot(clean_df['price'], bins=50, kde=True, color='teal')
plt.title('Cleaned Price Distribution ($10 to $1500)')
plt.show()

print(f"Original dataset size: {len(df)} rows")
print(f"Cleaned dataset size: {len(clean_df)} rows")
```

```python
# 1. Fill missing numerical values (like review scores) with the median
num_cols = clean_df.select_dtypes(include=['float64', 'int64']).columns
clean_df[num_cols] = clean_df[num_cols].fillna(clean_df[num_cols].median())

# 2. Fill missing categorical values (like host_response_time) with a placeholder
cat_cols = clean_df.select_dtypes(include=['object', 'string']).columns
clean_df[cat_cols] = clean_df[cat_cols].fillna('Unknown')

# 3. Verify no missing values remain (should output 0)
print(f"Maximum missing values in any column: {clean_df.isnull().sum().max()}")

# 4. Export the clean, processed data for the ML pipeline
processed_path = '../data/processed.csv'
clean_df.to_csv(processed_path, index=False)
print(f"Processed dataset saved to: {processed_path}")
```
